# Exportar al DW

```python
df_cliente_ml = (df_cliente
                 .merge(df_edad_cliente, on='id_tipo_cliente')
                 .merge(df_nivel_cliente, on='id_nivel_cliente')
                 )

columnas_id = [c for c in df_cliente_ml.columns if 'id_' in c.lower() and c not in 'id_cliente']

df_final_cliente = df_cliente_ml.drop(columns= columnas_id + ['requerimiento_anual','beneficio_descuento'] )

df_final_cliente.head(3)
```

```python
df_ml = (
    df_ventas
    .merge(df_productos, on='id_producto')
    .merge(df_tienda, on= 'id_tienda')
    .merge(df_final_cliente, on= 'id_cliente')
    )

columnas_id = [c for c in df_ml.columns if 'id_' in c.lower()]
df_final = df_ml.drop(columns=columnas_id)
```
